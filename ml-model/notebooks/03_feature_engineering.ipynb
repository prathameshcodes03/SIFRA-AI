```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/cleaned.csv")

df.head()
```

```python
print(df.columns)
```

```python
df["knee_angle_missing"] = df["knee_angle"].isnull().astype(int)

df["elbow_angle_missing"] = df["elbow_angle"].isnull().astype(int)
```

```python
df["knee_angle"] = df["knee_angle"].fillna(0)

df["elbow_angle"] = df["elbow_angle"].fillna(0)
```

```python
df = pd.get_dummies(
    df,
    columns=["exercise"],
    prefix="exercise",
    dtype=int
)

df.head()
```

```python
X = df.drop(columns=["form_label"])

y = df["form_label"]
```

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

y_encoded = label_encoder.fit_transform(y)
```

```python
print("X shape:", X.shape)
print("y shape:", y_encoded.shape)

print("\nMissing values in X:")
print(X.isnull().sum())
```

```python
X = X.astype(float)

print(X.dtypes)
```

```python
X.to_csv(
    "../data/sifra_ai_features.csv",
    index=False
)

pd.DataFrame({
    "form_label": y,
    "form_label_encoded": y_encoded
}).to_csv(
    "../data/sifra_ai_target.csv",
    index=False
)

print("Feature and target datasets saved successfully.")
```
