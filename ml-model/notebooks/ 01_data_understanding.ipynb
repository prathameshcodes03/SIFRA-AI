```python
df['knee_angle']=df['knee_angle'].fillna(df['knee_angle'].mean())
df['elbow_angle']=df['elbow_angle'].fillna(df['elbow_angle'].mean())
df['body_angle']=df['body_angle'].fillna(df['body_angle'].mean())
df['symmetry_score']=df['symmetry_score'].fillna(df['symmetry_score'].mean()) 
df['landmark_visibility']=df['landmark_visibility'].fillna(df['landmark_visibility'].mean()) 



df.isna().sum()
```

```python
df.dtypes
```

```python
# Numerical columns
numeric_cols = [
    "knee_angle",
    "elbow_angle",
    "hip_angle",
    "body_angle",
    "symmetry_score",
    "landmark_visibility"
]

# Convert numerical columns to numeric
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Categorical columns
df["exercise"] = df["exercise"].astype("string")
df["form_label"] = df["form_label"].astype("string")

# Check again
df.dtypes
```

```python
# Check angle ranges
angle_cols = [
    "knee_angle",
    "elbow_angle",
    "hip_angle",
    "body_angle"
]

for col in angle_cols:
    print(f"\n{col}")
    print(df[(df[col] < 0) | (df[col] > 180)][[col]])
```

```python
print("Exercise values:")
print(df["exercise"].unique())

print("\nForm labels:")
print(df["form_label"].unique())
```

```python
valid_labels = [
    "correct",
    "shallow",
    "forward_lean",
    "knees_caving",
    "pushup_correct",
    "pushup_elbow_flare",
    "pushup_body_sag",
    "pushup_hips_high",
    "plank_correct",
    "plank_hips_high",
    "plank_hips_low"
]

invalid_labels = df[~df["form_label"].isin(valid_labels)]

print("Invalid form-label rows:")
print(invalid_labels)
```
